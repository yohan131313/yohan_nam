# counter 클래스 만들기
- 데이터: 현재숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter :
    def __init__(self,initial_count=0) : #생성함수
        self.count = initial_count

    def increment(self, amount=1) :
        self.count += amount

    def get_count(self) :
        return self.count

def test_counter() :
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10) :
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())

if __name__ == "__main__" :
    test_counter()
```

### 클래스 circle
### 클래스 속성 -(프라이빗) radius 
### + get_radius: float


```python
import math
class Circle:
    def __init__(self,radius = 0):
        self.radius = radius
    def get_Area(self):
        return math.pi*self.radius*self.radius
    def getPerimeter(self):
        return 2*math.pi*self.radius
c = Circle(10)
print("원의면적", c.get_Area())
print("원의넓이", c.getPerimeter())
```

### person 객체 
* 클래스명 : person
* 속성: 이름, 출생년도
* 연산: 
 - 네 이름은?
 - 올해 네 나이는?
* 테스트 함수: test_person
 - 우리팀의 4명에 대한 객체를 생성하고 목록에 저장
 - for 문을 통해 목록에 있는 모든 객체의 이름과 니이 출력
* 메인함수
 - 테스트 함수 수행

```python
from datetime import datetime


class Person:
    def __init__(self, name, birth_year):
        self.__name = name
        self.__birth_year = birth_year

    def get_name(self):
        return self.__name

    def get_current_year(self):
        return datetime.now().year

    def get_age(self):
        current_year = self.get_current_year()
        return current_year - self.__birth_year

    def __str__(self):
        return f"Name: {self.__name}, Age: {self.get_age()}"


def test_person():
    team_member = [
        Person("Nam", 2003),
        Person("Jo", 2004),
        Person("Hyuang", 2007),
        Person("He", 2007)
    ]
    return team_member


if __name__ == "__main__":
    members = test_person()

    for member in members:
        print(member)
```
