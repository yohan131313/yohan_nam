### 객체지향의 3가지 핵심요소
1. 핵심화
데이터와 연산을 하나의 덩어리로 묶음
정보은닉을 위한 가시성: 
public 공개(모든 클래스에서 접근가능), protected 보호(자식만 접근가능), private 숨김(접근불가)

2. 상속
부모의 private 한 것은 상속은 받으나 접근 불가능
protected는 외부 접근 불가능하나 자식이 물려받음

3. 다형성
문맥에 따라 다른 함수가 수행됨
A.do(), B.do() 함수 이름은 같으나 클래스가 다르면 다른 함수

### 상속과 is-a 관계
ex) 푸들은 강아지다
ex) 자동차는 차량이다

```python
#상속 구현하기
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, make):
        self.maker = make

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker:{self.maker}, Model:{self.model}, Color:{self.color}, Price:{self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery:{self.batterySize}kwc"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, PayLoad:{self.payLoad}kg"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())


def main():
    myCar = ElectricCar("Tesla", "Model S", "white", 10000, 0)
    myCar.setMaker("Tesla")
    myCar.setBatterySize(60)

    print(myCar.getDesc())


main()
print(car1.model)
ecar1 = ElectricCar("Tesla","Model S","Red",8000,81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())
truck1 = Truck("Ford", "F-150", "Black", 4000, 1000)
print(truck1)
print(truck1.getMaker())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))
print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))
```
