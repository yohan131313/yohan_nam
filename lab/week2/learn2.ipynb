```python
#lab 사각형 그리기
import turtle
t = turtle.Turtle()
t.shape("turtle")
size = int(input("사각형의 크기는 얼마로 할까요? "))
t.forward(size)
t.right(90) 
t.forward(size)
t.right(90)
t.forward(size)
t.right(90)
t.forward(size)
turtle.done()
```

```python
#mini project 거북이를 조종하는 프로그램 만들기
import turtle

t = turtle.Turtle()
t.shape("turtle")

color = input("색상(blue, red, yellow만 가능): ")
t.color(color)

while True:
    print("\n거리와 각도를 입력하시오.")
    
    angle_input = input("거북이 회전 각도(left): ")
    if not angle_input:
        break
        
    distance_input = input("거북이 전진 거리: ")
    if not distance_input:
        break

    angle = float(angle_input)
    distance = float(distance_input)

    t.left(angle)
    t.forward(distance)

    print("제어에 성공하였습니다.")

turtle.done()
```

```python
#서로다른 인수로 호출가능
def get_area(radius:float) -> float:
    area = 3.14*radius**2
    return area
result1 = get_area(3)
result2 = get_area(20)
print("반지름이 3인 원의 면적=", result1)
print("반지름이 20인 원의 면적=", result2)
```

```python

def addsub(x, y):
    a = x + y
    b = x - y
    return a,b

x = 10
y = 4
a,b = addsub(x, y)

print(f"{x} + {y} = {a}")
print(f"{x} - {y} = {b}")
```
