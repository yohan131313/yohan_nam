### 오류의 종류

사용자 입력 오류: 사용자가 정수를 입력하여야 하는데 실수를 입력할 수 있다.
장치 오류: 네트워크가 안 된다거나 하드 디스크 작동이 실패할 수 있다.
코드 오류: 잘못된 인덱스를 사용하여서 배열에 접근할 수 있다.
  IOError: 파일을 열 수 없으면 발생한다.
  importError: 파이썬이 모듈을 찾을 수 없으면 발생한다.
  ValueError: 연산이나 내장 함수에서 인수가 적절치않은 값을 가지고 으면 발생한다.
  KeyboardInterrupt: 사용자가 인터럽트 키를 누르면 발생한다. (Control-C나 Delete)
  EOFError: 내장 함수가 파일의 끝을 만나면 발생한다


```python
# Lab: 매출 파일 처리

infilename = input("입력 파일 이름: ")
outfilename = input("출력 파일 이름: ")

try:
    infile = open(infilename, "r")
    outfile = open(outfilename, "w")
    
    total_sum = 0
    count = 0
    
    line = infile.readline()
    while line != "":
        s = int(line)
        total_sum += s
        count += 1
        line = infile.readline()
        
    outfile.write("총매출 = " + str(total_sum) + "\n")
    if count > 0:
        outfile.write("평균 일매출 = " + str(total_sum / count))
    else:
        outfile.write("평균 일매출 = 0")
        
    infile.close()
    outfile.close()
    print("처리가 완료되었습니다.")

except FileNotFoundError:
    print("지정한 파일을 찾을 수 없습니다.")
```

```python
import random

try:
    infile = open("words.txt", "r")
    lines = infile.readlines()
    infile.close()
    words = [line.strip() for line in lines if line.strip()]
except FileNotFoundError:
    words = ["python", "program", "computer", "hangman", "university"]

word = random.choice(words)
guesses = ''
turns = 10

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char, end="")
        else:
            print("_", end="")
            failed += 1
            
    print()
    
    if failed == 0:
        print("사용자가 이겼습니다!")
        break
        
    guess = input("글자를 추측하세요: ")
    guesses += guess
    
    if guess not in word:
        turns -= 1
        print(f"틀렸습니다! 남은 기회: {turns}")
        
        if turns == 0:
            print("사용자가 졌습니다.")
```

```python
# Lab: 각 문자 횟수 세기
filename = input("파일명을 입력하세요: ").strip()

try:
    infile = open(filename, "r")
    freqs = {}
    
    for line in infile:
        for char in line.strip():
            if char in freqs:
                freqs[char] += 1
            else:
                freqs[char] = 1
                
    print(freqs)
    infile.close()
    
except FileNotFoundError:
    print("지정한 파일을 찾을 수 없습니다.")
```

```python
# Lab: 파일 암호화
key = 'abcdefghijklmnopqrstuvwxyz'

def encrypt(n, plaintext):
    result = ''
    for l in plaintext.lower():
        try:
            i = (key.index(l) + n) % 26
            result += key[i]
        except ValueError:
            result += l
    return result.lower()

def decrypt(n, ciphertext):
    return encrypt(-n, ciphertext)

plain = "hello python"
encrypted_text = encrypt(3, plain)
decrypted_text = decrypt(3, encrypted_text)

print(f"평문: {plain}")
print(f"암호문: {encrypted_text}")
print(f"복호문: {decrypted_text}")
```

## 이진파일
0과 1로 구성
 순차접근
 임의접근: 파일 포인터는 읽기와 쓰기 동작이 현재 어떤 위치에서 이루어지는 지를 나타낸다, 임의로 파일포인터 위치를 바꿀 수 있다.

 



```python
# Lab: 이미지 파일 복사하기
filename1 = "123.png"
filename2 = "kkk.png"

infile = open(filename1, "rb")
outfile = open(filename2, "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()

print(filename1 + "를 " + filename2 + "로 복사하였습니다.")
```

 정규식
 정규식(regular expression)이란 특정한 규칙을 가지고 있는 문자열들을 메타 문자를 이용하여 표현하는 수식이다.
 re 모듈을 사용 : re.search()

 

```python
# Lab: 패스워드 검사 프로그램
import re

password = input("패스워드를 입력하세요: ")
flag = 0

while True:
    if len(password) < 8:
        flag = -1
        break
    elif not re.search("[a-z]", password):
        flag = -1
        break
    elif not re.search("[A-Z]", password):
        flag = -1
        break
    elif not re.search("[0-9]", password):
        flag = -1
        break
    elif not re.search("[_@$]", password):
        flag = -1
        break
    else:
        flag = 0
        print("유효한 패스워드")
        break

if flag == -1:
    print("유효한 패스워드가 아닙니다.")
```
