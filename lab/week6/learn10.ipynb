### 파일의 기초
프로그램에서 만든 데이터를 영구히 저장하고자 한다면 하드 디스크에 파일 형태로 저장하여야한다.
메모리 VS SSD,하드디스크

### 파일의 개념

첫번째 바이트의 위치를 오프셋이라고 하고 1씩증가
열린 파일 포인터는 최초에 파일을 열 때는 0을 가리키며 접근한 바이트 수만큼 증가한다

파일의 종류
 이진파일: 이진수의 바이트 배열이 저장된 ㅏ일
 텍스트 파일: 각 바이트가 문자로 구성된 파일

경로
