```python
# lab 학생 성적처리
score_dic = {
"Kim":[99,83,95],
"Lee":[68,45,78],
"Choi":[25,56,69]
}
for name, scores in score_dic.items():
    print(name,"의 평균성적=",sum(scores)/len(scores))
```

```python
# mini project 주소록 작성

address_book = {}

while True:
    print("1. 연락처 추가")
    print("2. 연락처 삭제")
    print("3. 연락처 검색")
    print("4. 연락처 출력")
    print("5. 종료")
    
    choice = input("메뉴 항목을 선택하시오: ")
    
    if choice == '1':
        name = input("이름: ")
        phone = input("전화번호: ")
        address_book[name] = phone
    elif choice == '2':
        name = input("삭제할 이름: ")
        if name in address_book:
            del address_book[name]
            print(f"{name} 님의 연락처가 삭제되었습니다.")
        else:
            print("존재하지 않는 이름입니다.")
    elif choice == '3':
        name = input("검색할 이름: ")
        if name in address_book:
            print(f"{name} 의 전화번호: {address_book[name]}")
        else:
            print("존재하지 않는 이름입니다.")
    elif choice == '4':
        if not address_book:
            print("저장된 연락처가 없습니다.")
        else:
            for name, phone in address_book.items():
                print(f"{name} 의 전화번호: {phone}")
    elif choice == '5':
        print("프로그램을 종료합니다.")
        break
    else:
        print("올바른 메뉴 번호를 입력해주세요.")
```
