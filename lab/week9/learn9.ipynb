### 기본 위젯(Widgets) 종류
Label : 텍스트나 이미지를 화면에 표시하는 읽기 전용 위젯
Button : 사용자가 클릭했을 때 특정 함수나 명령을 실행하는 
Entry : 한 줄의 텍스트를 입력받을 수 있는 입력창
Text : 여러 줄의 텍스트를 입력하거나 출력할 수 있는 편집기 형태의 위젯
Checkbutton : 여러 개 중 0개 이상을 중복해서 선택할 수 있는 체크박스
Radiobutton : 여러 개 중 단 하나의 옵션만 선택할 수 있는 라디오 버튼
Listbox : 여러 줄의 항목 목록을 보여주고 선택하게 하는 목록 상자
Combobox : ttk 모듈에 있는 항목 목록 드롭다운 선택창
Scale : 슬라이더를 마우스로 끌어 연속된 숫자 값을 선택하는 위젯
Scrollbar : 텍스트나 리스트박스, 캔버스 등에 연결해 화면을 스크롤하는 위젯
Canvas : 선, 도형, 이미지 등을 자유롭게 그리고 배치할 수 있는 도화지 영역
Frame : 다른 위젯들을 담는 컨테이너(구조물) 역할의 영역
Toplevel : 메인 창 외에 새로 생성되는 독립된 보조 창(팝업창)
Menu / Menubutton : 상단 메뉴바나 드롭다운 메뉴를 만드는 위젯 [1, 2]

## 컨테이너 종류
1. Fram프레임e () tk.Frame

2. LabelFrame (라벨 프레임) tk.LabelFrame

PanedWindow (페인드 윈도우) tk.PanedWindow
 내부에 배치된 위젯들 사이에 마우스로 드래그하여 크기를 조절할 수 있는 구분선(경계선)을 제공
5. Notebook (노트북 / 탭) ttk.Notebook
 웹 브라우저의 탭이나 책의 색인처럼 여러 개의 페이지(Tab)를 전환하며 볼 수 있게 해주는 컨테이너
6. Canvas (캔버스) tk.Canvas

테이너 역할도 겸합
7. Fram프레임e () tk.Frame
 복잡한 화면을 구역별로 나눌때
8. LabelFrame (라벨 프레임) tk.LabelFrame
 화면안에서 연관된 위젯들을 시각적으로 명확하게 한 그룹으로 묶을때
9. Toplevel tk.Toplevel

10. PanedWindow tk.PanedWindow

11. Notebook ttk.Notebook

### 압축 배치 관리자
위젯 (버튼, 레이블 등)을 위젯을 화면에 배치할 때 구체적인 좌표(X, Y)를 지정하지 않고, 마치 물건을 상자 안에 차곡차곡 압축하여 밀어 넣듯이 배치하는 방식
쌓는 방향 : side = TOP(“top”), BOTTOM(“bottom”), LEFT (“left”), RIGHT (“right”)
ex. window 안에 3개의 label 을 차곡차곡 세로로 쌓음 (기본 : side = TOP)

축 배치 관리자
위젯 (버튼, 레이블 등)을 위젯을 화면에 배치할 때 구체적인 좌표(X, Y)를 지정하지 않고, 마치 물건을 상자 안에 차곡차곡 압축하여 밀어 넣듯이 배치하는 방식
ex. window 안에 3개의 label 을 차곡차곡 가로로 쌓음 (side = LEFT : 왼쪽으로 팩킹)

격자 배치 관리자
위젯 (버튼, 레이블 등)을 테이블 형태로 배치
ex. 버튼1 : (0,0), 버튼 2 (0,1), 버튼 3(1,0), 버튼 4(1,1)

절대위치 배치 관리자
위젯 (버튼, 레이블 등)을 지정한 절대 좌표(x,y)에 배치

```python
import random
from tkinter import *

window = Tk()
window.title("가위바위보 게임")

Label(window, text="선택하세요", font=("Helvetica", "16")).pack()
frame = Frame(window)

def decide(player_choice):
    options = ["가위", "바위", "보"]
    computer_choice = random.choice(options)
    
    # 승패 판정
    if player_choice == computer_choice:
        result = "비겼습니다!"
    elif (
        (player_choice == "가위" and computer_choice == "보") or
        (player_choice == "바위" and computer_choice == "가위") or
        (player_choice == "보" and computer_choice == "바위")
    ):
        result = "이겼습니다!"
    else:
        result = "졌습니다ㅠ"
        
    output.config(text=f"내 선택: {player_choice} / 컴퓨터: {computer_choice}\n결과: {result}")

def pass_s():
    decide("가위")
def pass_r():
    decide("바위")
def pass_p():
    decide("보")
    
rock = Button(frame, text="✊ 바위", font=("Helvetica", "14"), width=10, command=pass_r)
rock.pack(side="left", padx=5)
paper = Button(frame, text="✋ 보", font=("Helvetica", "14"), width=10, command=pass_p)
paper.pack(side="left", padx=5)
scissors = Button(frame, text="✌️ 가위", font=("Helvetica", "14"), width=10, command=pass_s)
scissors.pack(side="left", padx=5)
frame.pack(pady=10)

output = Label(window, text="", font=("Helvetica", "16"))
output.pack(pady=10)

window.mainloop()
```

```python
# Lab: 계산기 프로그램
from tkinter import *

window = Tk()
window.title("간단한 계산기")

# 입력창 생성 (결과 및 입력값 표시)
entry = Entry(window, width=16, font=("Helvetica", 20), justify="right")
entry.grid(row=0, column=0, columnspan=4, padx=10, pady=10)

def btn_click(item):
    current = entry.get()
    entry.delete(0, END)
    entry.insert(0, str(current) + str(item))

def btn_clear():
    entry.delete(0, END)

def btn_equal():
    try:
        # 입력된 문자열을 파이썬 수식으로 계산
        result = str(eval(entry.get()))
        entry.delete(0, END)
        entry.insert(0, result)
    except Exception:
        entry.delete(0, END)
        entry.insert(0, "오류")

# 버튼 구성 요소 (텍스트, 행, 열)
buttons = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3)
]

# 반복문을 통한 버튼 생성 및 배치
for (text, row, col) in buttons:
    if text == '=':
        b = Button(window, text=text, width=5, height=2, font=("Helvetica", 14), command=btn_equal)
    else:
        b = Button(window, text=text, width=5, height=2, font=("Helvetica", 14), command=lambda t=text: btn_click(t))
    b.grid(row=row, column=col, padx=2, pady=2)

# 초기화(C) 버튼 추가
btn_c = Button(window, text="C", width=22, height=2, font=("Helvetica", 14), command=btn_clear)
btn_c.grid(row=5, column=0, columnspan=4, padx=2, pady=2)

window.mainloop()
```

```python
# Lab: 도형 그리기

from tkinter import *

WIDTH = 600
HEIGHT = 200

def displayRect():
    canvas.create_rectangle(10, 10, WIDTH-10, HEIGHT-10)

def displayOval():
    canvas.create_oval(10, 10, WIDTH-10, HEIGHT-10, fill="yellow")

def displayArc():
    canvas.create_arc(10, 10, WIDTH-10, HEIGHT-10, start=0, extent=120, width=10, fill='blue')

def displayPolygon():
    canvas.create_polygon(10, 10, WIDTH-10, HEIGHT-10, 200, 90, 300, 160)

def displayLine():
    canvas.create_line(10, 10, WIDTH-10, HEIGHT-10, fill='green')

def clearCanvas():
    canvas.delete("all")

window = Tk()
window.title("캔버스 도형 그리기")

canvas = Canvas(window, width=WIDTH, height=HEIGHT, bg='white')
canvas.pack()

frame = Frame(window)
frame.pack(pady=10)

btLine = Button(frame, text="Line", command=displayLine)
btLine.grid(row=1, column=1, padx=2)

btRectangle = Button(frame, text="Rectangle", command=displayRect)
btRectangle.grid(row=1, column=2, padx=2)

btOval = Button(frame, text="Oval", command=displayOval)
btOval.grid(row=1, column=3, padx=2)

btPolygon = Button(frame, text="Polygon", command=displayPolygon)
btPolygon.grid(row=1, column=4, padx=2)

btArc = Button(frame, text="Arc", command=displayArc)
btArc.grid(row=1, column=5, padx=2)

btClear = Button(frame, text="Clear", command=clearCanvas)
btClear.grid(row=1, column=6, padx=2)

window.mainloop()
```

```python
# Lab: 그림판 프로그램 만들기
from tkinter import *
from tkinter import colorchooser

class PaintApp:
    def __init__(self, root):
        self.root = root
        self.root.title("간단한 그림판")
        
        # 기본 설정 (펜 굵기, 색상, 지우개 모드)
        self.line_width = 2
        self.line_color = "black"
        self.eraser_on = False
        
        # 버튼을 담을 상단 프레임 생성
        frame = Frame(root)
        frame.pack(side=TOP, fill=X, padx=5, pady=5)
        
        # 제공해주신 버튼 코드 적용
        penButton = Button(frame, text='펜', command=self.use_pen)
        penButton.grid(row=0, column=0, sticky=W+E, padx=2)
        
        brushButton = Button(frame, text='브러쉬', command=self.use_brush)
        brushButton.grid(row=0, column=1, sticky=W+E, padx=2)
        
        colorButton = Button(frame, text='색상선택', command=self.choose_color)
        colorButton.grid(row=0, column=2, sticky=W+E, padx=2)
        
        eraseButton = Button(frame, text='지우개', command=self.use_eraser)
        eraseButton.grid(row=0, column=3, sticky=W+E, padx=2)
        
        clearButton = Button(frame, text='모두삭제', command=self.clear_all)
        clearButton.grid(row=0, column=4, sticky=W+E, padx=2)
        
        # 버튼들이 고르게 공간을 차지하도록 설정
        for i in range(5):
            frame.columnconfigure(i, weight=1)
        
        # 그림을 그릴 캔버스 생성
        self.canvas = Canvas(root, bg='white', width=600, height=400)
        self.canvas.pack(fill=BOTH, expand=True, padx=5, pady=5)
        
        # 마우스 이벤트 연결
        self.setup_variables()

    def setup_variables(self):
        self.old_x = None
        self.old_y = None
        self.canvas.bind('<Button-1>', self.reset)
        self.canvas.bind('<B1-Motion>', self.draw)

    def use_pen(self):
        self.line_width = 2
        self.eraser_on = False

    def use_brush(self):
        self.line_width = 10
        self.eraser_on = False

    def choose_color(self):
        # 색상 팔레트 창 띄우기
        color = colorchooser.askcolor(color=self.line_color)[1]
        if color:
            self.line_color = color
            self.eraser_on = False

    def use_eraser(self):
        self.eraser_on = True

    def clear_all(self):
        self.canvas.delete("all")

    def reset(self, event):
        self.old_x = event.x
        self.old_y = event.y

    def draw(self, event):
        # 지우개 모드일 경우 흰색으로 두껍게 칠하기
        if self.eraser_on:
            paint_color = 'white'
            paint_width = 20
        else:
            paint_color = self.line_color
            paint_width = self.line_width
            
        if self.old_x and self.old_y:
            self.canvas.create_line(
                self.old_x, self.old_y, event.x, event.y,
                width=paint_width, fill=paint_color,
                capstyle=ROUND, smooth=TRUE, splinesteps=36
            )
        self.old_x = event.x
        self.old_y = event.y

if __name__ == "__main__":
    window = Tk()
    app = PaintApp(window)
    window.mainloop()
```

```python
# lab 공 애니메이션 

from tkinter import *
import time
import random

window = Tk()   
window.title("통통 튀는 공 애니메이션")
canvas = Canvas(window, width=600, height=400, bg="white")
canvas.pack()

class Ball:
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)
        
    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        
        # 원이 위쪽이나 아래쪽 벽에 부딪혔을 때
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
            
        # 원이 왼쪽이나 오른쪽 벽에 부딪혔을 때
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

# 공 객체 생성
ball1 = Ball("blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

# 애니메이션 루프 실행
while True:
    ball1.move()
    ball2.move()
    ball3.move()
    window.update()
    time.sleep(0.05)
```

```python
#lab 공애니메이션2
from tkinter import *
import time
import random

window = Tk()
window.title("공 30개 튀기기 게임")
canvas = Canvas(window, width=600, height=400, bg="white")
canvas.pack()

class Ball:
    def __init__(self, color, size):
        # 겹치지 않게 무작위 시작 위치 지정
        start_x = random.randint(0, 400)
        start_y = random.randint(0, 300)
        self.id = canvas.create_oval(start_x, start_y, start_x + size, start_y + size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)
        
    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        
        # 위·아래 벽 충돌 시 부호 반전
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
            
        # 왼쪽·오른쪽 벽 충돌 시 부호 반전
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
ballList = []

# 공 30개 생성
for i in range(30):
    ballList.append(Ball(random.choice(colors), 40))

# 애니메이션 반복문
while True:
    for i in range(30):
        ballList[i].move()
    window.update()
    time.sleep(0.03)

window.mainloop()
```
