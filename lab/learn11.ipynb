```python
# Lab: 키를 이용한 정렬 예제

class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def __repr__(self):
        return "<이름: %s, 나이: %s>" % (self.name, self.age)
        
    def getName(self):
        return self.name

# 정렬 함수
def keyAge(person):  # 크기를 비교할 값 (나이)
    return person.age

people = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]

print(sorted(people, key=keyAge))
print(sorted(people, key=lambda x: x.getName()))
```

람다식
 람다식는 이름은 없고 몸체만 있는 함수이다.
 람다식은 lambda 키워드로 만들어진다.
 람다식은 딱 한 번 사용하는 함수를 만드는데 사용된다.

```python
# Lab: 람다식으로 온도 변환하기
f_temp = [0, 10, 20, 30, 40, 50]
c_temp = map(lambda x: (5.0/9.0)*(x-32.0), f_temp)
print(list(c_temp))
```

```python
# Lab: 람다식으로 데이터 처리하기
orders = [
    ["1", "재킷", 5, 12000],
    ["2", "셔츠", 6, 24000],
    ["3", "바지", 3, 51000],
    ["4", "코드", 6, 200000]
]

result = list(map(lambda x: (x[0], x[2] * x[3]), orders))
print(result)
```

이터레이터
 파이썬에서는 for 루프와 함께 사용할 수 있는 여러 종류의 객체가 있으며 이들 객체는 이터러블 객체
(iterable object)이라고 불린다.
 __iter__ ()은 이터러블 객체 자신을 반환한다.
 __next__ ()은 다음 반복을 위한 값을 반환한다. 만약 더 이상의 값이 없으면 StopIteration 예외를 발생
하면 된다.

제너레이터
 제너레이터(generators)는 키워드 yield를 사용하여서 함수로부터 이터레이터를 생성하는 하나의 방법
이다.


```python
# Lab: 피보나치 이터레이터

class FibIterator:
    def __init__(self, a=1, b=0, maxValue=50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration()
        self.a = self.b
        self.b = n
        return n

for i in FibIterator():
    print(i, end=" ")
```

연산자 오버로딩
 연산자 메소드로 정의하는것을 연산자 오버로딩이라고함


```python
# Lab: Book 클래스

class Book:
    def __init__(self, title='', pages=0):
        self.title = title
        self.pages = pages

    # Book 객체끼리의 + 연산 정의 (페이지 수 합산)
    def __add__(self, other):
        if isinstance(other, Book):
            return self.pages + other.pages
        return self.pages + other

    def __str__(self):
        return self.title

    def __gt__(self, other):
        return self.pages > other.pages

# --- 테스트 코드 ---
book1 = Book("파이썬 프로그래밍", 300)
book2 = Book("자료구조와 알고리즘", 450)

# 두 책의 페이지 수 더하기 (+ 연산자 사용)
total_pages = book1 + book2
print("두 책의 총 페이지 수:", total_pages)

# 페이지 수 비교 (> 연산자 사용)
print("book1이 book2보다 페이지가 더 많은가?", book1 > book2)
```

오버로딩 할 수 있는 연산자
 덧셈: `x.__add__(y)`
 뺄셈: `x.__sub__(y)`
 곱셈: `x.__mul__(y)`
 지수: `x.__pow__(y)`
 나눗셈(실수): `x.__truediv__(y)`
 나눗셈(정수): `x.__floordiv__(y)`
 나머지: `x.__mod__(y)`
 작음: `x.__lt__(y)`
 작거나 같음: `x.__le__(y)`
 같음: `x.__eq__(y)`
 같지 않음: `x.__ne__(y)`
 큼: `x.__gt__(y)`
 크거나 같음: `x.__ge__(y)`
